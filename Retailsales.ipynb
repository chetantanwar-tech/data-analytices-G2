```python
import pandas as pd
```

```python
df = pd.read_csv('sales_data.csv', encoding='latin-1')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
df['Postal_Code'] = df.groupby('Category')['Postal_Code'].transform(lambda x: x.fillna(x.median))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
```

```python
df.columns = df.columns.str.replace(' ', '_')
```

```python
df.columns
```

```python
df['postal_code'] = df['postal_code'].fillna(0)
```

```python
# convert dates
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)
df['ship_date'] = pd.to_datetime(df['ship_date'], dayfirst=True)
```

```python
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['day_of_week'] = df['order_date'].dt.day_name()

df['shipping_days'] = (df['ship_date'] - df['order_date']).dt.days
```

```python
#Validate data for quick sanity checks
df[['order_date', 'ship_date', 'shipping_days']].head()
df['shipping_days'].describe()
```

```python
df = df[df['shipping_days'] >= 0]
```

```python
df['sales'].sum()
df['sales'].mean()
```

```python
#categories make which money
df.groupby('category')['sales'].sum().sort_values(ascending=False)
```

```python
#best region for generating sales
df.groupby('region')['sales'].sum().sort_values(ascending=False)
```

```python
#sales over time
df.groupby('year')['sales'].sum()
df.groupby('month')['sales'].sum()
```

```python
# best products
df.groupby('sub_category')['sales'].sum().sort_values(ascending=False)
```

```python
# customer segments
df.groupby('segment')['sales'].sum()
```

```python
segment_sales = df.groupby('segment')['sales'].sum().sort_values(ascending=False)
```

```python
print(segment_sales)
```

```python
import matplotlib.pyplot as plt

category_sales = df.groupby('category')['sales'].sum().sort_values(ascending=False)

category_sales.plot(kind='bar')
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
region_sales = df.groupby('region')['sales'].sum().sort_values(ascending=False)

region_sales.plot(kind='bar')
plt.title("Sales by Region")
plt.show()
```

```python
monthly_sales = df.groupby('month')['sales'].sum()

monthly_sales.plot(kind='line', marker='o')
plt.title("Monthly Sales Trend")
plt.show()
```

```python
top_products = df.groupby('sub_category')['sales'].sum().sort_values(ascending=False).head(10)

top_products.plot(kind='bar')
plt.title("Top 10 Sub-Categories")
plt.show()
```

```python
segment_sales = df.groupby('segment')['sales'].sum()

segment_sales.plot(kind='bar')
plt.title("Sales by Segment")
plt.show()
```
